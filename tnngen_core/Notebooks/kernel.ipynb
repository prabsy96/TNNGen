```python
rfsize = 1
nprev = 1
inputsize = 9
stride = 1
num_k = int((((inputsize-rfsize)/stride)+1) * (((inputsize-rfsize)/stride)+1))
in_width = inputsize*inputsize*nprev
p = rfsize*rfsize*nprev
out_width = p*num_k
```

```python
layer_in = np.arange(in_width)
layer_out = np.zeros(out_width)
```

```python
n_col = int(((inputsize-rfsize)/stride)+1)
n_row = int(((inputsize-rfsize)/stride)+1)
col = inputsize
row = inputsize

for k in range(num_k):
    ky = int(np.floor(k/n_col))
    kx = int(np.floor(k%n_row))
    
    for r in range(rfsize):
        out1 = rfsize*rfsize*nprev*k+r*rfsize*nprev
        out2 = out1+(rfsize*nprev-1)
        in1 = ((kx + ky*inputsize)*stride + inputsize*r)*nprev
        in2 = in1+(rfsize*nprev-1)
        #print(((kx + ky*inputsize)*stride + inputsize*r))
        #print(kx*stride)
        #print(in1, in2)
        layer_out[out1:out2+1] = layer_in[in1:in2+1]
```
